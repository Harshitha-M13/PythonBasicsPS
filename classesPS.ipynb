# _ _init_ _ Method/Function

### Constructor function(__init__)
All the classes have a function called __init__(), which is always executed/invoked when the object is being initiated(started)


### self parameter:
The self parameter is a reference to the current instance/object of the class, and is used to access variables that belongs to the class

```python
class Student:
    name = "GowriShankar"

    def __init__(self):
        print(self)         # See both of it is same(location)
        print("Welcome!")

# stu_n.name is working because of "inheritance", Ex - 
"""
    1. stu_1(1st object) will lookup whether it has any name attribute/ variable
    2. If "No" then it will lookup in a class
    3. If it(stu_1) finds it it will return the value, else it will return an Error!
"""
stu_1 = Student()           # 1st object
print(stu_1)                # See both of it is same(location)
print(stu_1.name)

stu_2 = Student()           # 2nd object(observe different location)
print(stu_2)   
print(stu_2.name)  

stu_3 = Student()           # 3rd object(observe different location)
print(stu_3) 
print(stu_3.name)



# Compare
print(id(stu_1))
print(id(stu_2))
print(id(stu_3))
```

# Class and Instance Attributes

* Class.attr
* Obj.attr

```python
# also an example of method over-writing
class Student:
    clg_name = "BMS College for Women"      # class attribute - requires less memory (Because it is a blueprint and doesn't consume more memory)

    # Default constructor - no other parameter other than self
    def __init__(self):
        print("GowriShankar!")

    # Parameterized constructor - composed of other parameter other than self
    def __init__(self, name, marks):
        self.name = name                    # object attribute
        self.marks = marks
        print(f"Our name is {name} and marks is {marks}")
    
stu_1 = Student("GowriShankar", 99)       # Every time the "Student" class is called - it creates a object that requires more memory - Ex: attributes will be created everytime it forms an instance or object, So it is better to code in class if all the methods inside the class needs that attribute(in our case color)

# access college name
print(stu_1.clg_name)
print(Student.clg_name)     # class_name.attr

print(stu_1.name)
print(stu_1.marks)


name = Student()           
print(name)
```

# Methods

### Methods are functions that belongs to objects.

```python
class Student:
    clg_name = "BMS College for Women"

    def __init__(self, name, marks):
        print(self)             # same location - of object(same object)

        self.name = name
        self.marks = marks

    def welcomeing_stu(self):
        print(self)             # same location - of object(same object) - Since those  "self" is object we can extract object attributes

        print("Welcome", self.name)

    def get_marks(self):
        return self.marks
    
stu_1 = Student("GowriShankar", 99)
```

```python
# Create student class that takes name and marks of 3 students as arguments in constructor. Then create a method toprint the average

class Student:

    def __init__(self, marks_list):
        print("avg method calculates average of 3 marks.")
        self.marks = marks_list     # stays in the memory if we use "self" - doesn't get deleted

    def avg(self):
        total_marks = 0        # temporily created and will get deleted/ removed after the execution of - avg() method
        
        for mark in self.marks:
            total_marks += mark

        avg = total_marks / 3

        return avg
    

stu_1 = Student([99, 95, 93])
    
```

```python
# Solving it in still more simpler form

class Student:

    def __init__(self, name, marks_list):
        self.name = name
        self.marks = marks_list

    def avg(self):
        return sum(self.marks) / len(self.marks)
```

```python
class OurList:         # most preferable is CamelCase
    def __init__(self):     # cmd + shift + l
        """Initialize an empty list."""
        self._items = []

    def add(self, item):
        """Add an item to the end of the list."""
        self._items.append(item)

    def __getitem__(self, index):
        """Retrieve an item by index."""
        return self._items[index]
    
    def __setitem__(self, index, value):
        """Set an item at a specific index."""
        self._items[index] = value

    def __repr__(self):
        """Return a string representation of the list."""
        return str(self._items)
    def __len__(self):
        """Return the length of the list."""
        return len(self._items)
```

```python
class Cal:
    def __init__(self):
        self._items = []

    def __addEle__(self, ele):
        self._items.append(ele)
        print(self._items)
    
    def __getEle__(self, index):
        return self._items[index]
    
    def __setEle__(self, index, ele):
        self._items[index] = ele
        print(self._items)

    def __repr__(self):
        return str(self._items)
    
    def __len__(self):
        return len(self._items)
    
    def __rm__(self, index):
        self._items.remove(index)
        print(self._items)
    
```

```python
class CalculateSalary:
    def __init__(self, base_salary, bonus_rate = .1, symbol = "$"):
        self.symbol = symbol
        self.base_salary = base_salary
        self.bonus_rate = bonus_rate
        self.total_salary = base_salary * (1 + bonus_rate)
        self.bonus = base_salary * bonus_rate

    def __repr__(self):             # will Run by default - no need to "call"
        return f"Base Salary: {self.symbol}{self.base_salary:,.2f}"
    
    def show_total_salary(self):
        return f"Total Salary: {self.symbol}{self.total_salary:,}"
    
    def show_bonus(self):
        return f"Bonus: {self.symbol}{self.bonus:,.0f}"
    

salary = CalculateSalary(30_000)
```
