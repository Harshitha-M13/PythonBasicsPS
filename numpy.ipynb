```python
# Not a number - nan

# Job Titles 
job_titles = np.array(["Data Analyst", "Data Scientist", "Data Engineer", "Machine Learning Engineer", "Ai Engineer"])

# Base Salaries
Base_salaries = np.array([60_000, 80_000, 90_000, 85_000, np.nan])

# Bonus rates
Bonus_rates = np.array([.10, .15, .13, .16, np.nan])

# Total salaries
Total_salaries = Base_salaries * (1 + Bonus_rates)


Total_salaries
```

```python
np.mean(Total_salaries)         # Because, anything multiplied or calculations performed with "nan" is "nan" itself 
```

```python
np.nanmean(Total_salaries)
```
