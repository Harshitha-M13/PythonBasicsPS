```python
req_skill_for_job = ["Python", "SQL", "Tableau", "Ml", "Ai Ethics"]



job_description = {
    "job-seeker-1": ["Python", "SQL", "Excel", "Power BI"],
    "job-seeker-2": ["Python", "SQL", "Tableau", "Ml", "Ai Ethics"],
    "job-seeker-3": ["HTML", "CSS", "js", "MERN stack"] 
}


job_infos = [job_info for job_info in list(job_description.items())]    # List comprehension

print(job_infos)

for job_info in job_infos:
    for skill in job_info[1]:
        if skill in all(req_skill_for_job):     
            print(job_info[0], skill)

    
```
