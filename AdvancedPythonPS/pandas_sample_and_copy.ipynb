# sample() = randomly pick some data from your DataFrame.

* sample(n=3): If we want a specific number of random rows, then we should use n.
* sample(frac=0.4): frac means fraction. Instead of saying how many rows we want, we say what proportion of the DataFrame we want
 
 Example :
Our DataFrame has 5 rows.

df.sample(frac=0.4)

0.4 = 40%.

So:

40% of 5 rows = 2 rows

Therefore, it randomly selects approximately 2 rows.

* df.sample(n=2, random_state=42): Using random_state=42 ensures that you get the same sample, 42 isn't special. we could use - random_state=10, random_state=100
* df.sample(n=3, replace=True): Normally, sample() selects rows without replacement, when set to "True" it means "Duplicates are allowed"
* df.sample(n=2, axis=1): Sampling columns, By default, sample() selects rows. "This randomly selects 2 columns"




df.sample(

    frac=None,

    replace=False,

    random_state=42,

# copy()

```python
import pandas as pd
from datasets import load_dataset

df = load_dataset("lukebarousse/data_jobs")["train"].to_pandas()


df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df.info()
```

```python
df_assigned = df
```

```python
df_assigned.loc[:5, "salary_year_avg"]
```

```python
df_assigned.loc[:5, ["salary_year_avg"]]
```

```python
median_salary_year_avg = df["salary_year_avg"].median()
```

```python
median_salary_year_avg
```

```python
df_assigned["salary_year_avg"] = df_assigned["salary_year_avg"].fillna(median_salary_year_avg)
```

```python
df_assigned.loc[:5, ["salary_year_avg"]]
```

```python
df.loc[:5, ["salary_year_avg"]]
```

```python
id(df) == id(df_assigned)
```

```python
# Using copy() method from pandas

df_copy = df.copy()
```

```python
print(f"ID of df:                                                           {id(df)}")
print(f"ID of df_copy:                                                      {id(df_copy)}")
print(f"Are the two dataframe pointing towards the same object ?            {id(df) == id(df_copy)}")
```
