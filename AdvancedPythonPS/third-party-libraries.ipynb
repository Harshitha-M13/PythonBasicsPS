# not vs ~ 

## Suppose we have one light bulb.

### Not

#### works
* On -> Off 

* Off -> On


#### doesn't work (Error!) - "not" does not work all together only supported for single value
* On    ->      Off
* Off   ->      On
* On    ->      Off 
* On    ->      Off
* Off   ->      On
* On    ->      Off
  

### ~
#### works all together
* On    ->      Off
* Off   ->      On
* On    ->      Off 
* On    ->      Off
* Off   ->      On
* On    ->      Off


```python
# To extract the True - "is null"

df[not df['total_bedrooms'].isnull()]     # Beacuse it consists of many Boolean values  
```

```python
# To extract the False - "is null"

df[~ df['total_bedrooms'].isnull()]   

# or 

df[df["total_bedrooms"].notnull()]

# or

df[df["total_bedrooms"].notna()]
```
