# Tuples

### Immutable

```python
tup_1 = (1, 2, 3)
print(id(tup_1))        # Proof that the variable is not the same - old internal reference

tup_2 = (3, 4, 5, 6)

tup_1 += tup_2          # Observe it is being assigned then why is it called immutable? - because it changes it internal reference to new one and that old variable remains

print(tup_1)

print(id(tup_1))        # Proof that it is now pointing to another "new" internal reference (same thing happens to int class also)
```

```python
# But this does't happen in list or dictionary

list_1 = [1, 2, 3]
list_2 = [3, 4, 5]
print(id(list_1))       # Observe list_1 id

list_1 += list_2

print(list_1)

print(id(list_1))       # Observe the id of list_1 once again, the changes as been made in the original place
```

# Identity operators

### Used to check if two values share the same memory location

* is
* is not

```python
emp1_salary = 10_000    #python interpreter ignores "_"
emp2_salary = 10_000

print(emp1_salary == emp2_salary)       # Compares the value
print(emp1_salary is emp2_salary)       # compares the memory address - id()
print(emp1_salary is not emp2_salary)   # compares the memory address - id()

emp2_salary = emp1_salary

print(emp1_salary is emp2_salary)       # Because - (id(emp1_salary) == id(emp2_salary))
print(emp1_salary is not emp2_salary)   # Because - (id(emp1_salary) != id(emp2_salary))
```

```python
# Dictionary - { role: skill }

job_description = {
    "Data Scientist": ["Python", "SQL", "Tableau", "Ml", "Ai Ethics"],
    "Data Analysts": ["Python", "SQL", "Excel", "Power BI"],
    "Web-developer": ["HTML", "CSS", "js", "MERN stack"] 
}

job_description
```

```python
for job_info in job_description:        # By default they are set to .keys()
    print(job_info)

# or - provides same results

for job_info in job_description.keys():        
    print(job_info)
```

```python
# If we want to access the values

for job_info in job_description.values():        
    print(job_info)

# or

for job_info in job_description:
    print(job_description[job_info])
```

```python
for job_info in job_description.items():        
    print(job_info)
```

```python
for job_info in job_description.items():        
    print(job_info[0], job_info[1])

print("\n")

# or

for job_title, skills in job_description.items():        
    print(job_title, skills)
```

```python
# while loop - Display skills required for a Data Scientist, "For loops would be best, but this is just for practice and learning"
i = 0
job_info = list(job_description.items())
print(f"job_info = {job_info} \n")


while job_info[i][0] in "Data Scientist":
    print(job_info[i][1])
    i += 1

```

```python
i = 0
len_dic = len(job_description)      # 3

while i < len_dic:

    job_title, skills_req = list(job_description.items())[i]

    if(job_title in "Data Scientist"):
        print(f"Skills required for {job_title} is {skills_req}")
    i += 1

```

```python
# Display the name of the employess who have more than 5 years of experience

years_of_exp = {
    "SitaRam": 9,
    "RadhaKrishn": 10,
    "SidhiBudhiGanu": 3
}

i = 0
len_emp = len(years_of_exp)
emp_info = list(years_of_exp.items())

while i < len_emp:

    emp_name, emp_exp = emp_info[i]

    if(emp_exp > 5):
        print(emp_name, "\n")
    
    i += 1
```

# List Comprehension

```python
# normal loop

our_list = []

for number in range(1, 11):
    our_list.append(number)


print(our_list)

# or

# Using list comprehension
our_list = [number for number in range(1, 11)]

print(our_list)
```

```python
# Using if statements in list comprehension

# Syntax - [var_name for var_name in (iterable) if(condition)]      var_name = is iterator

# Dictionary - { role: skill }

job_description = {
    "Data Scientist": ["Python", "SQL", "Tableau", "Ml", "Ai Ethics"],
    "Data Analysts": ["Python", "SQL", "Excel", "Power BI"],
    "Web-developer": ["HTML", "CSS", "js", "MERN stack"] 
}

skills_DS = [skills_req for job_title, skills_req in list(job_description.items()) if job_title in "Data Scientist"]

print(skills_DS)

print(skills_DS[0])
```

```python
is_python = [job_title for job_title, skills_req in list(job_description.items()) if "Python" in skills_req]

print(is_python)

print(len(is_python))           # Length of list

print(is_python[0])

print(is_python[1])
```
