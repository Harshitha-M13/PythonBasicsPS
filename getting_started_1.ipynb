```python
#New Income Tax slab rates
"""
Upto 4,00,000 = NIL
4,00,001 - 8,00,000 = 5%
8,00,001 - 12,00,000 = 10%
12,00,001 - 16,00,000 = 15%
16,00,001 - 20,00,000 = 20%
20,00,001 - 24,00,000 = 25%
above 24,00,000 = 30% 
"""

income = int(input("Enter the Income(LPA): "))
tax = 0

if(income <= 400000):
    tax = 0
elif(income > 400000 and income <= 800000):
    tax = 0.5
elif(income > 800000 and income <= 1200000):
    tax = 0.10
elif(income > 1200000 and income <= 1600000):
    tax = 0.15
elif(income > 1600000 and income <= 2000000):
    tax = 0.20
elif(income > 2000000 and income <= 240000):
    tax = 0.25
else:
    tax = 0.30

tax_payable = income * tax

print(f"Since your income is: {income}")
print(f"Tax rate = {tax * 100}%")
print(f"Total tax payable is : {tax_payable}")
    
```

# Attribute: the variable of an Object
### Attribute: A Field in a record, defined by it's class

```python
#Best Practice
class Cars:
    # By default python will call this __init__(which is a constructor function), This is what happening internally 
    def __init__(self): 
        self.color = "blue"     #Here color is an attribute
        self.brand = "Polo"     #brand is also an attribute

car1 = Cars()

print(car1.color)
print(car1.brand)
```

```python
class Student:
    #Parameterized Constructor
    def __init__(self, name, percentage, grade):
        self.name = name
        self.percentage = percentage
        self.grade = grade


stu_1 = Student("GowriShankar", 93.5, "A")
print(stu_1.name)
print(stu_1.percentage)
print(stu_1.grade)
```
