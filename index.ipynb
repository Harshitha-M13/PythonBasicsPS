```python
import copy as cp

dad = "dad"
correct_ans = "dad"

reversed_str = ""

for ch in reversed(dad):
    reversed_str += ch


if reversed_str == correct_ans:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
# Shallow Copy → Copies only the first (outer) level. Nested objects are shared.
# Deep Copy → Copies everything, including all nested objects, so the original and copy are completely independent.

import copy

a = [1, [2, 3], 4]

b = copy.copy(a)
c = copy.deepcopy(a)

b[1][0] = 100
c[1][1] = 200

print(a)
print(b)
print(c)
```

```python
all_emp_skills = {"Python", "SQL", "Tableau", "js", "R", "java", "html"}
emp1_skills = {"Python", "SQL"}
emp2_skills = {"Python", "SQL", "Tableau", "js", "R", "java", "html", "jax"}

all_emp_skills >= emp1_skills       # True
emp1_skills >= all_emp_skills       # False, because emp1_skills is not superset(i.e - everything in all_emp_skills(subset) is not inside/included in emp1_skills)

all_emp_skills > emp2_skills        # False, though all_emp_skills is superset and is having elements greater than emp2_skills(subset), but it doesn't consists all the elements emp2_skills(subset) - "Swift is not in superset(all_emp_skills)"
all_emp_skills < emp2_skills        # False

```

```python
our_skills = ["Python", "SQL", "Machine Learning", "CV", "Excel", "Tableau"]

job_infos = [
    {
        "role": "Data Scientist",
        "skills": ["Python", "SQL", "Tableau", "Machine Learning", "Ai Ethics", "Excel"]
    }, 
    {
        "role": "Data Analysts",
        "skills": ["Python", "SQL", "Tableau", "Excel"]
    },
    {
        "role": "Machine Learning engineer",
        "skills": ["Python", "TensorFlow", "PyTorch", "Scikit-Learn"]
    }
]
```

```python
for job_info.items() in job_infos:      # for job_info - it is already in list and .items() is the method of dict
    print(job_info)
```

```python
for job_info in job_infos:
    print(job_info.items())
```

```python
for job_info in job_infos:
    job_details = list(job_info.items())
    role, skills = job_details[0][1], job_details[1][1]

    print(role, skills)
```

```python
qualified_roles = [job_info["role"] for job_info in job_infos if all([skill in our_skills for skill in job_info["skills"]])]
print(qualified_roles)
```
