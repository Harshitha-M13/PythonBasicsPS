# Sets - unordered collections of unique items

### sets are mutable - because we can use &=, |=, remove(), update() 
### No indexing because - unordered collection
### But it's elements are immutable, because - sets stores only string, int, tuples(immutable), and not list, dictionary(mutable), for example:


#### It includes 
* {1, 2, 3}             
* {"apple", "orange", "Mango"}
* {(1, 2), (3, 4)}

#### It doesn't include
* {[1, 2], [3, 4]}
* {
    "key": value
}


```python
# set

our_set = {"Python", "SQL", "Tableau", "Python", "R", "java", "Javascript"}

print(our_set)
```

```python
our_set[0]  # WE can't access element by index, because Sets are unordered collections of unique items - TypeError: 'set' object is not subscriptable
```

```python
# But if we want to access these set elements, we can just convert theses sets into lists or tuple

# For example 
"Set -> list"
skills_list = list(our_set)[0]

print(skills_list)


"list -> set - for analysisng the uniques elements in the set(for data analysis and understanding)"
skills_list = ["Python", "SQL", "Python", "Excel", "R", "Tableau", "Python"]

set(skills_list)
```

```python
set_1 = {1, 2, 3, 4}
set_2 = {3, 4, 5, 6}

ele_present_in_both = set_1.__and__(set_2)      # common from both sets
print(ele_present_in_both)

set_1 & set_2
```

```python
i_and = set_1.__iand__(set_2)

print(i_and)        # it is similar to .__and__ but the original set changes
print(set_1) 

print(set_1)
print(set_2)

set_1 &= set_2      # Take common from both and assign the result back to set_1 - {3, 4}
print(set_1)


set_2 &= set_1      # Take common from both and assign the result back to set_2 - {3, 4}
print(set_2)
```

```python
all_emp_skills = {"Python", "SQL", "Tableau", "js", "R", "java", "html"}
emp1_skills = {"Python", "SQL"}
emp2_skills = {"Python", "Swift", "R", "SQL", "Tableau", "html"}

all_emp_skills >= emp1_skills       # True
emp1_skills >= all_emp_skills       # False, because emp1_skills is not superset(i.e - everything in all_emp_skills(subset) is not inside/included in emp1_skills)

all_emp_skills > emp2_skills        # False, though all_emp_skills is superset and is having elements greater than emp2_skills(subset), but it doesn't consists all the elements emp2_skills(subset) - "Swift is not in superset(all_emp_skills)"
all_emp_skills < emp2_skills        # False
```
