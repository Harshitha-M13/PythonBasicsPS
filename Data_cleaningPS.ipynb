```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading Datasets 
df = load_dataset("lukebarousse/data_jobs")["train"].to_pandas()

# Data cleanup
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

# Using .fillna()

```python
median_salary_year = df["salary_year_avg"].median()

median_salary_year
```

```python
median_salary_hour = df["salary_hour_avg"].median()

median_salary_hour
```

```python
df[["salary_year_avg", "salary_hour_avg"]].median()
```

```python
df_fillna_with_median = df  # Is pointing to the same address in memory
```

```python
id(df) == id(df_fillna_with_median)
```

```python
df_fillna_with_median["salary_year_avg"].fillna(value=median_salary_year, inplace=True) # It cannot guarantee that this temporary Series is the original DataFrame's actual column. - that's why we get "ChainedAssignmentError"
```

```python
df_fillna_with_median.loc[:10, :]       # No replacement is done
```

```python
# We can do this instead, because we are re-assigning it
df_fillna_with_median["salary_year_avg"] = df_fillna_with_median["salary_year_avg"].fillna(value=median_salary_year)
```

```python
df.head(3)          # It has also effected in - the original DataFrame
```

```python
df_fillna_with_median.head(3)       # observe "salary_year_avg" column
```
