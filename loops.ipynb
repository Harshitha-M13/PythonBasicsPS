```python
# can also call the function in this way - Immediately Invoked Function Expression (IIFE)
'''
    Why is it called an Immediately Invoked Function Expression (IIFE)?

    Because the function is:

    - Created
    - Called immediately
    - Never stored in a variable

    It exists only for that one expression.
'''
(lambda x: x + 1)(2)
```

```python
mul_two = lambda x, y: x * y

mul_two(3, 3)
```

```python
# Real use-case of these lambda function

'Example  - calculating salary for the existing employees'

emp_salary_list = [50_000, 80_000, 90_000, 1_00_000, 5_00_000]

def calculate_salary(base_salary, bonus_rate = .1):

    total_salary = base_salary * (1 + bonus_rate)

    return total_salary

total_salary_cal = [calculate_salary(salary) for salary in emp_salary_list]

print(total_salary_cal)
```

```python
# Using lambda functions

emp_salary_list = [50_000, 80_000, 90_000, 1_00_000, 5_00_000]

total_salary_cal = [(lambda salary: salary * (1 + .1)) for salary in emp_salary_list]

print(total_salary_cal)
```

```python
# Using lambda functions

emp_salary_list = [50_000, 80_000, 90_000, 1_00_000, 5_00_000]

total_salary_cal = [(lambda salary: salary * (1 + .1))(salary) for salary in emp_salary_list]       # IIFE

print(total_salary_cal)
```

```python
# Using lambda functions - Best Practice because we are not storing the func, we are immediately calling it and executing - we don't want store the func

emp_salary_list = [50_000, 80_000, 90_000, 1_00_000, 5_00_000]

total_salary_cal = [salary * (1 + .1) for salary in emp_salary_list]       # IIFE

print(total_salary_cal)
```
